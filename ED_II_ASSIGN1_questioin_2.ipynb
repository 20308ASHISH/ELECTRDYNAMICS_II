```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-5, 5, 100)
t = np.linspace(-5, 5, 100)
V = [0.01,0.5,0.9999]

x_axis = np.zeros_like(x)
t_axis = np.zeros_like(t)
```

```python
for v in V:
  gamma = (1-v**2)**0.5
  x_new = (x - v*0)/gamma
  x_new1 = + v*t/gamma

  t_new = (t-v*0)/gamma
  t_new1 = + v*x/gamma

  plt.plot(x_new,x_new1, color = 'r', label="x' axis")
  plt.plot(t_new1,t_new, color = 'b', label="t' axis")
  plt.plot(x, x_axis, 'k--', label="x-axis (S)")
  plt.plot(t_axis, t, 'k--', label="t-axis (S)")

  plt.plot(x, x, 'g:', label="light cone (t=±x)")
  plt.plot(x, -x, 'g:')
  plt.legend()
  plt.grid(True)
  plt.axis('equal')
  plt.title('Lorentz Transformation of Axes (Boost along x) with' + str(v) + ' c')
  plt.show()
```
